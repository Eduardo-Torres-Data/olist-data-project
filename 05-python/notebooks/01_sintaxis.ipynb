```python
productos = [
    {'nombre': 'sofa', 'precio': 899.90, 'categoria': 'moveis', 'stock': 5},
    {'nombre': 'celular', 'precio': 1299.00, 'categoria': 'eletronicos', 'stock': 0},
    {'nombre': 'shampoo', 'precio': 25.50, 'categoria': 'beleza_saude', 'stock': 120},
    {'nombre': 'notebook', 'precio': 3200.00, 'categoria': 'eletronicos', 'stock': 3},
    {'nombre': 'mesa', 'precio': 450.00, 'categoria': 'moveis', 'stock': 0},
]

for i in productos:
    if i['stock'] == 0:
        print(f"Producto {i['nombre']} sin stock")

total = 0.0
for p in productos:
    total += p['precio']
print(f"El total de precios de productos es: {total:.2f}")

contador_stock = {}
for a in productos:
    if a['categoria'] in contador_stock:
        contador_stock[a['categoria']] += 1
    else:
        contador_stock[a['categoria']] = 1
print(contador_stock)

for b in productos:
    if b['stock'] == 0:
        print(f"{b['nombre']} Reponer urgente")
    elif b['stock'] <= 5:
        print(f"{b['nombre']} stock bajo")
    else:
        print(f"{b['nombre']} stock OK")


c = 0
while c < len(productos):
    if productos[c]['categoria'] == 'eletronicos':
        print(f"{productos[c]['nombre']} - {productos[c]['precio']:.2f}")
    c += 1
```

```python
temperaturas = [22, 19, 25, 30, 18, 27, 31, 15]

clasificacion = []
for t in temperaturas:
    if t < 18:
        clasificacion.append("Frío")
    elif 18 <= t <= 25:
        clasificacion.append("Templado")
    else:
        clasificacion.append("Caliente")
print(clasificacion)

contador = {}
for t in clasificacion:
    if t in contador:
        contador[t] += 1
    else:
        contador[t] = 1
print(contador)


categoria_recurrente = ""
contador_maximo = 0

for c in contador:
    if contador[c] > contador_maximo:
        contador_maximo = contador[c]
        categoria_recurrente = c

print(f"La categoría más recurrente es: {categoria_recurrente} con {contador_maximo} apariciones.")
```

```python
empleados = [
    {'nombre': 'carlos', 'salario': 850.0, 'departamento': 'ventas', 'activo': True},
    {'nombre': 'maria', 'salario': 1200.0, 'departamento': 'ti', 'activo': True},
    {'nombre': 'jose', 'salario': 700.0, 'departamento': 'ventas', 'activo': False},
    {'nombre': 'lucia', 'salario': 1500.0, 'departamento': 'ti', 'activo': True},
    {'nombre': 'pedro', 'salario': 600.0, 'departamento': 'ventas', 'activo': True},
    {'nombre': 'ines', 'salario': 2000.0, 'departamento': 'gerencia', 'activo': True},
]

for i in empleados:
    if i['salario'] < 800:
        print(f"{i['nombre']}: Aumento del 10%")
    elif i['salario'] <= 1500:
        print(f"{i['nombre']}: Aumento del 5%")
    else:
        print(f"{i['nombre']}: Sin aumento")

ventas = 0.0
ti = 0.0
gerencia = 0.0
for d in empleados:
    if d['activo'] == True:
        if d['departamento'] == 'ventas':
            ventas += d['salario']
        elif d['departamento'] == 'ti':
            ti += d['salario']
        elif d['departamento'] == 'gerencia':
            gerencia += d['salario']
print(f"Total ventas: {ventas:.2f}")
print(f"Total ti: {ti:.2f}")
print(f"Total gerencia: {gerencia:.2f}")
```
